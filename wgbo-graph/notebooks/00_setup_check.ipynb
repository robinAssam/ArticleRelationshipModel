# 00 — Setup check

Run this notebook first to confirm the environment is configured.
It loads the starter annotation file and prints the atoms.

```python
# Make the project's `src/` importable from notebooks
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print('Project root:', PROJECT_ROOT)
```

```python
from src.annotation.schema import load_atoms

atoms = load_atoms(PROJECT_ROOT / 'data' / 'annotations' / '7_454.json')
print(f'Loaded {len(atoms)} atoms for article 7:454')
for a in atoms:
    print(f'  {a.atom_id}  [{a.condition_type:14s}]  {a.text[:60]}...')
```

```python
# Quick: which fields are populated for each atom?
import pandas as pd

rows = []
for a in atoms:
    rows.append({
        'atom_id': a.atom_id,
        'type': a.condition_type,
        'actors': len(a.actors),
        'acts': len(a.acts),
        'temporal': len(a.temporal),
        'refs': len(a.explicit_references),
        'hierarchies': len(a.hierarchies),
        'residual': len(a.residual),
    })

pd.DataFrame(rows)
```
